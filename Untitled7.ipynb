```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)

# Plot settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Load the dataset

df = pd.read_csv('/content/city_day.csv')

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

```python
# Dataset information

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Convert Date column to datetime

df['Date'] = pd.to_datetime(df['Date'])

# Create useful date-related columns

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()

print(df[['Date', 'Year', 'Month', 'Month_Name']].head())
```

## Plot 1: Number of Records by City

This bar plot shows the number of air-quality observations recorded
for each city. It helps identify the cities with more or fewer observations.

```python
plt.figure(figsize=(14, 7))

city_counts = df['City'].value_counts()

sns.barplot(
    x=city_counts.index,
    y=city_counts.values
)

plt.title('Number of Records by City')
plt.xlabel('City')
plt.ylabel('Number of Records')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

## Plot 2: AQI Distribution

This histogram shows the distribution of AQI values in the dataset.
It helps understand the common AQI range and the presence of extreme values.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='AQI',
    bins=50,
    kde=True
)

plt.title('Distribution of AQI')
plt.xlabel('AQI')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## Plot 3: AQI Category Distribution

This count plot shows the number of observations belonging to each
AQI category and provides an overall view of air-quality conditions.

```python
plt.figure(figsize=(10, 6))

order = [
    'Good',
    'Satisfactory',
    'Moderate',
    'Poor',
    'Very Poor',
    'Severe'
]

sns.countplot(
    data=df,
    x='AQI_Bucket',
    order=order
)

plt.title('AQI Category Distribution')
plt.xlabel('AQI Category')
plt.ylabel('Number of Observations')
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

## Plot 4: Average AQI by City

This bar plot compares the average AQI of different cities.
It helps identify cities with relatively higher or lower air pollution.

```python
plt.figure(figsize=(14, 7))

avg_aqi_city = (
    df.groupby('City')['AQI']
    .mean()
    .sort_values(ascending=False)
)

sns.barplot(
    x=avg_aqi_city.values,
    y=avg_aqi_city.index
)

plt.title('Average AQI by City')
plt.xlabel('Average AQI')
plt.ylabel('City')

plt.tight_layout()
plt.show()
```

## Plot 28: Correlation Heatmap of Air Quality Variables

This heatmap displays the correlation between AQI and different pollutants.
It helps identify strong and weak relationships between numerical variables.

```python
pollutants = [
    'PM2.5',
    'PM10',
    'NO',
    'NO2',
    'NOx',
    'NH3',
    'CO',
    'SO2',
    'O3',
    'Benzene',
    'Toluene',
    'Xylene',
    'AQI'
]

corr = df[pollutants].corr()

plt.figure(figsize=(14, 11))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Air Quality Variables')

plt.tight_layout()
plt.show()
```

## Plot 29: Percentage Distribution of AQI Categories

This pie chart shows the percentage contribution of each AQI category.

```python
aqi_counts = df['AQI_Bucket'].value_counts().reindex(order).dropna()

plt.figure(figsize=(9, 9))

plt.pie(
    aqi_counts.values,
    labels=aqi_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage Distribution of AQI Categories')

plt.show()
```

## Plot 30: City-wise AQI Category Distribution

This heatmap shows the number of observations belonging to each AQI category
for every city. It helps compare air-quality conditions across cities.

```python
city_category = pd.crosstab(
    df['City'],
    df['AQI_Bucket']
)

# Arrange AQI categories in logical order
city_category = city_category.reindex(
    columns=[
        'Good',
        'Satisfactory',
        'Moderate',
        'Poor',
        'Very Poor',
        'Severe'
    ],
    fill_value=0
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    city_category,
    annot=True,
    fmt='g',
    cmap='YlOrRd',
    linewidths=0.5
)

plt.title('City-wise AQI Category Distribution')
plt.xlabel('AQI Category')
plt.ylabel('City')

plt.tight_layout()
plt.show()
```
